```python
ImagesArray = []
W,H = 100 ,120
for img in os.listdir(ImagesPath):
  img = cv2.imread(os.path.join(ImagesPath,img) ,cv2.IMREAD_GRAYSCALE)
  img = cv2.resize(img, (H,W))
  img = img.reshape(H,W,1)
  ImagesArray.append(img)
len(ImagesArray)
```

```python
y = [np.random.choice([0,1]) for i in range(len(ImagesArray))]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(np.array(ImagesArray), np.array(y), test_size=0.25, random_state=44, shuffle =True)

print('X_train shape is ' , X_train.shape)
print('X_test shape is ' , X_test.shape)
print('y_train shape is ' , y_train.shape)
print('y_test shape is ' , y_test.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
KerasModel = keras.models.Sequential([
        keras.layers.Conv2D(filters = 32, kernel_size = (3,3),strides = (1,1), padding = 'VALID' ,input_shape = (H,W,1)),# SAME
        keras.layers.MaxPooling2D(pool_size = (2,2), strides = (2,2)),
        keras.layers.BatchNormalization(),
        keras.layers.Conv2D(filters = 32, kernel_size = (4,4),strides = (1,1), padding = 'VALID'  ),
        keras.layers.MaxPooling2D(pool_size = (2,2), strides = (2,2)),
        keras.layers.BatchNormalization(),
        keras.layers.Flatten(),
        keras.layers.Dense(128, activation = tf.nn.relu),
        keras.layers.Dropout(0.2),
        keras.layers.Dense(1, activation = tf.nn.sigmoid),
        ])
```

```python
KerasModel.compile(optimizer ='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
KerasModel.fit(X_train, y_train, epochs=10)
```

```python
print(KerasModel.summary())
```

```python
ageList = ['(0-2)', '(4-6)', '(8-12)', '(15-20)',
		'(25-32)', '(38-43)', '(48-53)', '(60-100)']
```

```python
face_detector = dlib.get_frontal_face_detector()
```

```python
faces = face_detector(img)
```

```python
# Test on an Image
img = cv2.imread(os.path.join(ImagesPath, Images[0]))
frame = img.copy()
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
faces = face_detector(gray)
```

```python
from google.colab.patches import cv2_imshow
if not faces:
    mssg = 'No face detected'
    cv2.putText(img, f'{mssg}', (40, 40), cv2.FONT_HERSHEY_SIMPLEX, 2, (200), 2)
    cv2_imshow(img)
else:
    for face in faces:
        x, y, x2, y2 = face.left(), face.top(), face.right(), face.bottom()
        face_img = gray[y:y2, x:x2]

        # Preprocess the face for CNN
        face_img = cv2.resize(face_img, (H, W))  # Resize to model input
        face_img = face_img.reshape(1, H, W, 1) / 255.0  # Normalize and reshape

        # Predict Age
        age_preds = KerasModel.predict(face_img)
        age = ageList[np.argmax(age_preds)]  # Map prediction to age range

        # Draw bounding box and age on the image
        cv2.rectangle(frame, (x, y), (x2, y2), (0, 200, 200), 2)
        cv2.putText(frame, f'{age}', (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 255), 2)

    # Display the results
    cv2_imshow(frame)
```
